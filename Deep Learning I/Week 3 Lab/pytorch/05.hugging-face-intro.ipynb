4. Loading a Pre-trained Model
- The transformers library makes it simple to load pre-trained models.
- You can load models for specific tasks like text classification, text generation, or question answering.

```python
from transformers import pipeline

# Create a text classification pipeline
classifier = pipeline('sentiment-analysis')

# Test it on a sample sentence
result = classifier("I am forced to use Hugging Face transformers because everyone else uses it!")
print(result)
```

5. Model Architectures: What Are Transformers?
- Transformers are a type of neural network architecture designed for sequence-to-sequence tasks, such as language translation.
- They are based on the concept of attention mechanisms, which allow the model to weigh the importance of different parts of the input.

Some popular transformer architectures available in Hugging Face:
- BERT (Bidirectional Encoder Representations from Transformers): Good for understanding language (e.g., text classification, named entity recognition).
- GPT (Generative Pre-trained Transformer): Great for text generation tasks.
- T5 (Text-To-Text Transfer Transformer): Versatile model that can perform many tasks by converting all problems into text-to-text format.

```python
from transformers import AutoModelForSequenceClassification, AutoTokenizer


model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# Tokenize a sentence
inputs = tokenizer("I am learning NLP with Hugging Face!", return_tensors="pt")

# Run the model
outputs = model(**inputs)
```

6. Hugging Face Pipelines
- Pipelines are high-level interfaces that let you perform various NLP tasks using a pre-trained model without needing to manually define the model, tokenizer, or configuration

```python
# Load the GPT-2 model for text generation
generator = pipeline('text-generation', model='gpt2')

# Generate text based on a prompt
result = generator("Why does deep learning cost so much", max_length=50, num_return_sequences=5)
print(result)
```

7. Tokenization: Preparing Input for Models
Before feeding text into a transformer model, it needs to be tokenized. Tokenization involves breaking down text into smaller components like words or subwords, which the model can understand.

Example of tokenizing with a BERT tokenizer:

```python
from transformers import BertTokenizer

# Load a pre-trained tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Tokenize a sentence
tokens = tokenizer("Hello, Hugging Face!")
print(tokens)
```

8. Embeddings: Representing Text as Numbers

```python
from transformers import BertTokenizer, BertModel
import torch

# Load a pre-trained tokenizer and model
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')

# Tokenize a sentence
sentence = "Hello, Hugging Face!"
inputs = tokenizer(sentence, return_tensors="pt", padding=True, truncation=True)

# Forward pass through the model
with torch.no_grad():
    outputs = model(**inputs)

# Extract the embeddings
# `last_hidden_state` contains embeddings for each token
token_embeddings = outputs.last_hidden_state  # Shape: (batch_size, seq_len, hidden_size)

# To get a single vector for the entire sentence, you can use the [CLS] token embedding:
cls_embedding = token_embeddings[:, 0, :]  # Shape: (batch_size, hidden_size)

print("Shape of CLS embedding:", cls_embedding.shape)
```

# Hugging Face datasets Library


1. What are Hugging Face datasets?
    - The datasets library by Hugging Face is a lightweight and efficient library designed to handle large-scale datasets for machine learning, especially for NLP tasks. It supports:
    - Loading and processing datasets from the Hugging Face Hub (which contains thousands of datasets).
    - Efficient storage and processing for both small and large datasets.
    - Streaming for large datasets that don't fit into memory.
    - Dataset transformations for preprocessing and tokenization.

    It’s built to work seamlessly with the transformers library but can be used independently for any machine learning project.

3. Loading Datasets from the Hugging Face Hub
- The Hugging Face Hub contains thousands of datasets for a wide range of tasks, such as sentiment analysis, text classification, machine translation, and more.
- You can load datasets directly from the hub using the datasets library.


```python
from datasets import load_dataset

# Load the IMDB dataset for sentiment analysis
dataset = load_dataset('imdb')

# Display the structure of the dataset
print(dataset)
```

4. Exploring a Dataset
    - Once you’ve loaded a dataset, you can explore it to understand its structure and contents.



```python
# Inspect the dataset
print(dataset['train'].features)

# Display the first few rows of the training set
print(dataset['train'][0:3])

# Display basic statistics
print(dataset['train'].info)
```

5. Accessing Dataset Splits
Most datasets come with predefined splits, such as train, validation, and test. You can access these splits easily:

```python
# Access the training set
train_dataset = dataset['train']

# Access the test set
test_dataset = dataset['test']

# Check the size of each split
print(f"Training set size: {len(train_dataset)}")
print(f"Test set size: {len(test_dataset)}")
```

6. Preprocessing Datasets
- Before feeding data into models, especially transformers, you need to preprocess the data, which includes tasks like tokenization and cleaning.
- Let’s tokenize the text in a dataset using a tokenizer from the Hugging Face transformers library:

```python
from transformers import BertTokenizer

# Load a pre-trained BERT tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

# Define a function to tokenize text
def tokenize_function(examples):
    return tokenizer(examples['text'], padding="max_length", truncation=True)

# Apply tokenization to the dataset
tokenized_datasets = dataset.map(tokenize_function, batched=True)

# Inspect the tokenized data
print(tokenized_datasets['train'][0])
```

7. Creating a Custom Dataset
- You can also create custom datasets using a Python dictionary or a Pandas DataFrame and convert them into the datasets format.
- Creating a dataset from a dictionary:

```python
from datasets import Dataset

# Create a simple dataset with two fields: 'text' and 'label'
data = {
    'text': ['I love Hugging Face!', 'Transformers are awesome!', 'Datasets library is super helpful!'],
    'label': [1, 1, 1]
}

# Convert the dictionary into a Hugging Face dataset
custom_dataset = Dataset.from_dict(data)

# Display the dataset
print(custom_dataset)
```

Creating a dataset from a Pandas DataFrame:


```python
import pandas as pd
from datasets import Dataset

# Create a Pandas DataFrame
df = pd.DataFrame({
    'text': ['I love Hugging Face!', 'Transformers are awesome!', 'Datasets library is super helpful!'],
    'label': [1, 1, 1]
})

# Convert the DataFrame to a Hugging Face dataset
custom_dataset = Dataset.from_pandas(df)

# Display the dataset
print(custom_dataset)
```

8. Dataset Transformations
- The datasets library allows you to apply various transformations such as filtering, shuffling, and splitting your dataset.
- Filter Dataset:
    - You can filter rows based on a condition. For example, let's filter out negative reviews from the IMDB dataset:

```python
# Filter the dataset for only positive reviews (label == 1)
positive_reviews = dataset['train'].filter(lambda example: example['label'] == 1)

# Display the first positive review
print(positive_reviews[0])
```

Shuffle Dataset:
You can shuffle the dataset to randomize the order of the data:

```python
# Shuffle the dataset
shuffled_dataset = dataset['train'].shuffle(seed=42)

# Display the first few rows of the shuffled dataset
print(shuffled_dataset[0:3])
```

Split Dataset:
If your dataset doesn’t come with predefined splits, you can split it manually:

```python
# Split the dataset into training and test sets (80/20 split)
train_test_split = custom_dataset.train_test_split(test_size=0.2)

# Access the new splits
train_dataset = train_test_split['train']
test_dataset = train_test_split['test']
```
