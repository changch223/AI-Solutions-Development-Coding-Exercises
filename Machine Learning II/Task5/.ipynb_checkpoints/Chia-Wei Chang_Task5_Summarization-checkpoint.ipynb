# Summarization
## This notebook outlines the concepts behind Text Summarization

## Summarization
- concept of capturing very important gist of a long piece of text

### Types of Summarization
- 1. **Extractive Summarization**
    - Select sentences from the corpus that best represent the text
    - Arrange them to form a summary
- 2. **Abstractive Summarization**
    - Captures the very important sentences from the text
    - Paraphrases them to form a summary

## Task: Take a piece of text from wiki page and summarize them using Sumy
### Steps
- Install the necessary libraries
- Import the libraries
- Scrape the text from a pre-defined webpage
- Summarize

### Install Sumy

```python
#! pip install sumy
```

### Import the libraries
- HtmlParser
- Tokenizer
- TextRankSummarizer

```python
import requests
from bs4 import BeautifulSoup
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer

# Import the summarizers from sumy
from sumy.summarizers.text_rank import TextRankSummarizer
from sumy.summarizers.lex_rank import LexRankSummarizer
from sumy.summarizers.luhn import LuhnSummarizer
from sumy.summarizers.lsa import LsaSummarizer
```

### Scrape the text

```python
url = "https://en.wikipedia.org/wiki/Automatic_summarization"
```

```python
# Fetch the page content
response = requests.get(url)
html_content = response.content

# Parse HTML using BeautifulSoup
soup = BeautifulSoup(html_content, 'html.parser')

# Extract text from the main content area (Wikipedia articles are typically in <div class="mw-parser-output">)
content_div = soup.find("div", {"class": "mw-parser-output"})
paragraphs = content_div.find_all("p")

# Combine paragraph texts, filtering out empty ones
scraped_text = "\n".join(p.get_text() for p in paragraphs if p.get_text(strip=True))
```

### Summarize - TextRankSummarizer

```python
parser = PlaintextParser.from_string(scraped_text, Tokenizer("english"))

# Define the number of sentences in each summary
num_sentences = 5
```

```python
textrank_summarizer = TextRankSummarizer()
```

### Try different Summarizers
- LexRankSummarizer
- LuhnSummarizer
- LsaSummarizer

```python
lexrank_summarizer = LexRankSummarizer()
luhn_summarizer = LuhnSummarizer()
lsa_summarizer = LsaSummarizer()
```

### Import the summarizers

```python
textrank_summary = textrank_summarizer(parser.document, num_sentences)
lexrank_summary = lexrank_summarizer(parser.document, num_sentences)
luhn_summary = luhn_summarizer(parser.document, num_sentences)
lsa_summary = lsa_summarizer(parser.document, num_sentences)
```

### Create Summarizers

```python
def print_summary(title, summary):
    print(title)
    for sentence in summary:
        print(f"- {sentence}")
    print("\n")
```

```python
print_summary("TextRank Summary:", textrank_summary)
```

### LexRankSummarizer

```python
print_summary("LexRank Summary:", lexrank_summary)
```

### LuhnSummarizer

```python
print_summary("Luhn Summary:", luhn_summary)
```

### LsaSummarizer

```python
print_summary("LSA Summary:", lsa_summary)
```

### Import the library

```python
from gensim.summarization import summarize as gensim_summarize
```

### Scrape the text
- Use beautifulSoup to extract text (from Task1 of ML-1)

```python
url = "https://en.wikipedia.org/wiki/Automatic_summarization"
```

```python
# Fetch the page content
response = requests.get(url)
html_content = response.content

# Parse HTML using BeautifulSoup
soup = BeautifulSoup(html_content, 'html.parser')

# Extract text from the main content area (Wikipedia articles are typically in <div class="mw-parser-output">)
content_div = soup.find("div", {"class": "mw-parser-output"})
paragraphs = content_div.find_all("p")

# Combine paragraph texts, filtering out empty ones
scraped_text = "\n".join(p.get_text() for p in paragraphs if p.get_text(strip=True))
```

```python
scraped_text
```

```python
gensim_summary = gensim_summarize(scraped_text, ratio=0.2)
```

### Summarize
- **word_count**: maximum amount of words we want in the summary
- **ratio**: fraction of sentences in the original text should be returned as output

```python
print("\nBest Summary Gensim:")
print(gensim_summary)
```
