### Import the libraries

```python
from sklearn.metrics import precision_recall_fscore_support
from sklearn.naive_bayes import MultinomialNB
```

### Create Training phrases
...... create more like these for your own domain chatbot
- intent: [user utterances]

```python
training_phrases = {
    "help-me": ' '.join(["I have a problem",
                         "Hey i need some answers",
                         "Can you help me with this?",
                         "I need help",
                         "Please help me"
                        ]),
    "depression-problem": ' '.join(["I am feeling very sad",
                                    "I am depressed",
                                    "I feel lonely",
                                    "I don't have any friends",
                                    "I am always alone",
                                    "Why am I so sad all the time"
                                   ]),
    "alcohol-addiction": ' '.join(["I drink too much",
                                   "I am addicted to alcohol",
                                   "I can't stop drinking",
                                   "I need alcohol daily",
                                   "I think I'm an alcoholic",
                                   "I take wine everyday"
                                  ])
}
```

```python
training_phrases
```

### Collect all training documents as user utterances

```python
training_documents = list(training_phrases.values())
labels = list(training_phrases.keys())
```

```python
training_documents
```

### Tokenize the user utterances

```python
import nltk
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
stop_words = set(stopwords.words('english')) 

word_tokens = []
for sent in training_documents:
    word_tokens.append(word_tokenize(sent))

print(word_tokens)

filtered_sentence = [w for w in word_tokens[0] if not w in stop_words]

print(word_tokens)
print(filtered_sentence)
```

### Create Bag of Words Model
- Import CountVectorizer
- fit_transform on training documents
- get_feature_names()

```python
from sklearn.feature_extraction.text import CountVectorizer

# Create and fit the vectorizer
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(training_documents)

# Print the sparse matrix
print(X)

# Get feature names (vocabulary)
print(vectorizer.get_feature_names_out())
```

### Create a Classifier
- Import
- Create
- Train

```python
# Create and train the classifier
classifier = MultinomialNB()
classifier.fit(X, labels)

print("Classifier trained successfully!")
```

### Predict on user utterances
- create a user query
- Transform the query using vectorizer
- Run prediction on the transformed query

```python
# Make predictions
test_query = ["I am feeling very depressed"]
test_query_features = vectorizer.transform(test_query)
prediction = classifier.predict(test_query_features)
print(f"Predicted intent: {prediction[0]}")
```

### Putting it all together inside predict( )

```python
def predict(raw_queries):
    # Transform the raw queries using the same vectorizer
    queries = vectorizer.transform(raw_queries)
    
    # Return predictions using the trained classifier
    return classifier.predict(queries)
```

```python
predicted = predict(["I am very much sad", "can we talk?", "Can you help me?", "i take wine everyday and i cant live without it"])
expected = ["depression-problem", "help-me", "help-me", "alcohol-addiction"]
```

```python
predicted
```

```python
evaluation = precision_recall_fscore_support(expected, predicted)
evaluation
```

```python
metrics = {}
(metrics['p'], metrics['r'], metrics['f1'], _) = evaluation
metrics
```

### Returning the Answer

```python
responses = {
    "help-me": "I'm here to help. What's on your mind?",
    "depression-problem": "I understand you're going through a difficult time. Let's talk about it.",
    "alcohol-addiction": "I hear your concerns about alcohol. There are people and resources available to help.",
    "alcohol-response": "Good to know that. Now I can work with you in making you better."
}
```

```python
responses['alcohol-response']
```

```python
predicted = predict(["i take wine everyday and i cant live without it"])
# expected = ["alcohol-addiction"]
predicted
```

```python
def send_response(raw_queries):
    predicted = predict(raw_queries)
    intent = predicted[0]
    print(f"Detected intent: {intent}")
    return responses[intent]
```

```python
bot_response = send_response(["i take wine everyday and i cant live without it"])
print(bot_response)
```

### Typos - Edit distance, Phonetics

```python
tokens = ['problem', 'Hey', 'need', 'some', 'answers', 'help', 'me', 'with', 'this', 'Please', 'addicted', 'alcohol', 'love', 'daily', 'alcoholic', 'depressed', 'lonely', 'dont', 'friends', 'alone', 'always', 'sad', 'Why', 'all', 'time', 'Hi', 'Hey', 'there', 'Hola', 'Hi', 'How', 'are', 'you', 'doing', '?']
```

```python
from difflib import get_close_matches
def spell_checker(token):
    # Get close matches from our vocabulary
    matches = get_close_matches(token, tokens, n=1, cutoff=0.8)
    
    if matches:
        # Return the closest match and True flag if found
        return matches[0], True
    else:
        # Return original token and False flag if no match found
        return token, False
```

```python
corrected_word, flag = spell_checker('alcholic')
corrected_word
```

### Testing the Chatbot with Different Inputs

```python
# Test cases for different intents
test_cases = [
    # Help intent tests
    "Can someone help me?",
    "I need assistance",
    "Please help",
    
    # Depression intent tests
    "I've been feeling down lately",
    "Everything feels hopeless",
    "I'm so lonely and sad",
    
    # Alcohol addiction intent tests
    "I can't stop drinking",
    "I think I might be addicted to alcohol",
    "I need wine every single day"
]

print("Testing chatbot responses...\n")
for test_input in test_cases:
    print(f"User: {test_input}")
    response = send_response([test_input])
    print(f"Bot: {response}\n")
```
