### Import the libraries

```python
from sklearn.metrics import precision_recall_fscore_support
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import precision_recall_fscore_support
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer
```

### Create Training phrases
...... create more like these for your own domain chatbot
- intent: [user utterances]

```python
#training_phrases = {
#    "help-me": ' '.join(["I have a problem",
#                         "Hey i need some answers",
#                         "Can you help me with this?",
#                         "I need help",
#                         "Please help me"
#                        ]),
#}
```

```python
training_phrases = {
    "account-inquiry": [
        "How do I check my account balance?",
        "I need to know my current balance.",
        "Can you tell me my available funds?",
        "How can I check my transaction history?",
        "I want to see my recent transactions."
    ],
    
    "loan-inquiry": [
        "How can I apply for a loan?",
        "What are the interest rates for personal loans?",
        "Can I check my loan eligibility?",
        "I need details about mortgage options.",
        "What is the loan repayment process?"
    ],
    
    "credit-card-inquiry": [
        "How do I apply for a credit card?",
        "What are the benefits of your credit cards?",
        "I need to check my credit card balance.",
        "Can you tell me my credit limit?",
        "How do I pay my credit card bill?"
    ],
    
    "fund-transfer": [
        "How do I transfer money to another account?",
        "Can I send money internationally?",
        "What are the charges for wire transfers?",
        "How long does a bank transfer take?",
        "I need to make an urgent transfer."
    ],
    
    "fraud-reporting": [
        "I think there's unauthorized activity on my account.",
        "My credit card was used without my permission.",
        "How do I report fraud?",
        "Someone accessed my account illegally.",
        "I lost my debit card, how do I block it?"
    ],
    
    "bank-branch-inquiry": [
        "Where is the nearest bank branch?",
        "What are the bank's opening hours?",
        "How do I contact customer support?",
        "Do you have a branch near me?",
        "Can I book an appointment at the bank?"
    ]
}
```

```python
# Prepare training data
training_documents = []
labels = []
```

```python
for label, phrases in training_phrases.items():
    for phrase in phrases:
        training_documents.append(phrase)
        labels.append(label)
```

```python
if not training_documents:
    raise ValueError("Training documents are empty!")
```

```python
vectorizer = CountVectorizer(stop_words=None, token_pattern=r"(?u)\b\w+\b")
X_train = vectorizer.fit_transform(training_documents)
y_train = np.array(labels)
```

```python
training_phrases
```

### Collect all training documents as user utterances

```python
training_documents = list(training_phrases.values())
labels = list(training_phrases.keys())
```

```python
training_documents
```

### Create Bag of Words Model
- Import CountVectorizer
- fit_transform on training documents
- get_feature_names()

```python
# Make a bag of words model
```

```python
for label, phrases in training_phrases.items():
    for phrase in phrases:
        training_documents.append(phrase)
        labels.append(label)
```

```python
# Check if training_documents is not empty
if not training_documents:
    raise ValueError("Training documents are empty!")
```

### Create a Classifier
- Import
- Create
- Train

```python
classifier = MultinomialNB()
classifier.fit(X_train, y_train)
```

### Predict on user utterances
- create a user query
- Transform the query using vectorizer
- Run prediction on the transformed query

### Putting it all together inside predict( )

```python
def predict_user_query(user_query):
    X_test = vectorizer.transform([user_query])  # Transform query
    prediction = classifier.predict(X_test)
    return prediction[0]
```

```python
user_query = "account balance?"
predicted_label = predict_user_query(user_query)
```

```python
print(f"User Query: {user_query}")
print(f"Predicted Intent: {predicted_label}")
print("Feature names:", vectorizer.get_feature_names_out())
```

```python
test_queries = [
    "How can I check my account balance?",  # Account Inquiry
    "I want to see my recent transactions.",  # Account Inquiry
    "What are the interest rates for personal loans?",  # Loan Inquiry
    "Can I check my loan eligibility?",  # Loan Inquiry
    "How do I apply for a credit card?",  # Credit Card Inquiry
    "How do I pay my credit card bill?",  # Credit Card Inquiry
    "How do I transfer money to another account?",  # Fund Transfer
    "How long does a bank transfer take?",  # Fund Transfer
    "My credit card was used without my permission.",  # Fraud Reporting
    "I lost my debit card, how do I block it?",  # Fraud Reporting
    "Where is the nearest bank branch?",  # Bank Branch Inquiry
    "How do I contact customer support?",  # Bank Branch Inquiry
]

expected_labels = [
    "account-inquiry",
    "account-inquiry",
    "loan-inquiry",
    "loan-inquiry",
    "credit-card-inquiry",
    "credit-card-inquiry",
    "fund-transfer",
    "fund-transfer",
    "fraud-reporting",
    "fraud-reporting",
    "bank-branch-inquiry",
    "bank-branch-inquiry"
]
```

```python
def predict_queries(user_queries):
    X_test = vectorizer.transform(user_queries)  # Transform list of queries
    predictions = classifier.predict(X_test)  # Predict labels
    return predictions
```

```python
predicted_labels = predict_queries(test_queries)
```

```python
for query, predicted, expected in zip(test_queries, predicted_labels, expected_labels):
    print(f"Query: {query}")
    print(f"Predicted Intent: {predicted}")
    print(f"Expected Intent: {expected}")
    print("-" * 40)
```
