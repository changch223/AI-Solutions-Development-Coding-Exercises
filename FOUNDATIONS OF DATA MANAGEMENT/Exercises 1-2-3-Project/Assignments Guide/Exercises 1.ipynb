```python
# Define a function to check if there are rows with the same country, city, and state/region, 
# but with a non-empty postal code
def check_postal_code(row, df):
    if pd.isna(row["Postal Code"]):
        match = df[
            (df["Country/Region"] == row["Country/Region"]) &
            (df["City"] == row["City"]) &
            (df["Region"] == row["Region"]) &
            (df["State"] == row["State"]) &
            df["Postal Code"].notna()
        ]
        return True if not match.empty else False
    return None  # No need to check for rows with a non-empty postal code

# Apply the function to the dataframe
df["Postal Code Check"] = df.apply(lambda row: check_postal_code(row, df), axis=1)

# Filter out rows where the Postal Code is missing and display the check results
missing_postal_df = df[df["Postal Code"].isna()][["Country/Region", "City", "State", "Region", "Postal Code", "Postal Code Check"]]

print(missing_postal_df)
```
