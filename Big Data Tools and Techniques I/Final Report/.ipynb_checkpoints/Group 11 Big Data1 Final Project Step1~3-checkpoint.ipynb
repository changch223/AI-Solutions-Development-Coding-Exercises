# 1. Dataset Selection and Description

💾 Dataset Description:
Data Source: MovieLens (provided by GroupLens)
- This data has been cleaned up - users
who had less than 20 ratings or did not have complete demographic
information were removed from this data set

1. Data Format:
- The *.dat files are plain text files, with :: (double colons) as the delimiter.

2. Structured Data:

- movies.dat: MovieID::Title::Genres

- ratings.dat: UserID::MovieID::Rating::Timestamp

- users.dat: UserID::Gender::Age::Occupation::Zip-code

3. Sample Business Questions:

- Which movies are the most popular among different age groups?

- Do male and female users have different rating preferences?

- Which movie genres have the highest average ratings?

- Movies rated per user occupation?

# 2. Data Storage and Preprocessing (MongoDB)

## Change .dat → .json

### movies.dat → movies.json

```python
import json

with open("ml-1m/movies.dat", encoding='latin1') as f:
    data = []
    for line in f:
        movie_id, title, genres = line.strip().split("::")
        data.append({
            "MovieID": int(movie_id),
            "Title": title,
            "Genres": genres.split("|")
        })

with open("movies.json", "w", encoding='utf-8') as f_out:
    json.dump(data, f_out, indent=2, ensure_ascii=False)
```

### ratings.dat → ratings.json

```python
import json

ratings = []
with open("ml-1m/ratings.dat", encoding='latin1') as f:
    for line in f:
        user_id, movie_id, rating, timestamp = line.strip().split("::")
        ratings.append({
            "UserID": int(user_id),
            "MovieID": int(movie_id),
            "Rating": float(rating),
            "Timestamp": int(timestamp)
        })

with open("ratings.json", "w", encoding='utf-8') as f_out:
    json.dump(ratings, f_out, indent=2, ensure_ascii=False)
```

### users.dat → users.json

```python
users = []
with open("ml-1m/users.dat", encoding='latin1') as f:
    for line in f:
        user_id, gender, age, occupation, zip_code = line.strip().split("::")
        users.append({
            "UserID": int(user_id),
            "Gender": gender,
            "Age": int(age),
            "Occupation": int(occupation),
            "ZipCode": zip_code
        })

with open("users.json", "w", encoding='utf-8') as f_out:
    json.dump(users, f_out, indent=2, ensure_ascii=False)
```

```python
# 1. read movies.dat
movies = []
with open("ml-1m/movies.dat", encoding='latin1') as f:
    for line in f:
        movie_id, title, genres = line.strip().split("::")
        movies.append({
            "MovieID": int(movie_id),
            "Title": title,
            "Genres": genres.split("|")
        })

# 2. read ratings.dat
ratings = []
with open("ml-1m/ratings.dat", encoding='latin1') as f:
    for line in f:
        user_id, movie_id, rating, timestamp = line.strip().split("::")
        ratings.append({
            "UserID": int(user_id),
            "MovieID": int(movie_id),
            "Rating": float(rating),
            "Timestamp": int(timestamp)
        })

# 3. read users.dat
users = []
with open("ml-1m/users.dat", encoding='latin1') as f:
    for line in f:
        user_id, gender, age, occupation, zip_code = line.strip().split("::")
        users.append({
            "UserID": int(user_id),
            "Gender": gender,
            "Age": int(age),
            "Occupation": int(occupation),
            "ZipCode": zip_code
        })

# 4. read MongoDB
db.movies.drop()
db.ratings.drop()
db.users.drop()

db.movies.insert_many(movies)
db.ratings.insert_many(ratings)
db.users.insert_many(users)

print("✅ Successful import MongoDB movielens database！")
```

# Data Analysis（Use MongoDB）

```python
from collections import defaultdict
import statistics
```

```python
# Query the total number of documents in each collection
print("📽️ movies:", db.movies.count_documents({}))
print("⭐ ratings:", db.ratings.count_documents({}))
print("👤 users:", db.users.count_documents({}))
```

```python
# Limit to the sample 10000 samples to avoid timeout

ratings_sample = list(db.ratings.aggregate([
    { "$sample": { "size": 10000 } }
]))
users = {user["UserID"]: user for user in db.users.find()}
movies = {movie["MovieID"]: movie for movie in db.movies.find()}
```

### The most rated movie by each age group (based on MovieID)sample 10000

```python
age_movie_count = defaultdict(lambda: defaultdict(int))

for r in ratings_sample:
    user = users.get(r["UserID"])
    if not user:
        continue
    age = user["Age"]
    age_movie_count[age][r["MovieID"]] += 1

for age, movie_counts in age_movie_count.items():
    top_movie = max(movie_counts.items(), key=lambda x: x[1])
    movie_title = movies.get(top_movie[0], {}).get("Title", "Unknown")
    print(f"Age {age} → Top Movie: {movie_title} (Count: {top_movie[1]})")
```

### Average rating for each movie genre sample 10000

```python
genre_ratings = defaultdict(list)

for r in ratings_sample:
    movie = movies.get(r["MovieID"])
    if not movie:
        continue
    genres = movie["Genres"]
    for g in genres:
        genre_ratings[g].append(r["Rating"])

for genre, ratings in genre_ratings.items():
    avg = round(statistics.mean(ratings), 2)
    print(f"Genre: {genre} → Avg Rating: {avg}")
```

### Top 10 highest-rated movies (with at least 50 ratings) all sample

```python
pipeline_top10_movies = [
    {
        "$group": {
            "_id": "$MovieID",
            "avgRating": { "$avg": "$Rating" },
            "count": { "$sum": 1 }
        }
    },
    { "$match": { "count": { "$gte": 50 } } },
    { "$sort": { "avgRating": -1 } },
    { "$limit": 10 },
    {
        "$lookup": {
            "from": "movies",
            "localField": "_id",
            "foreignField": "MovieID",
            "as": "movie"
        }
    },
    { "$unwind": "$movie" },
    {
        "$project": {
            "Title": "$movie.Title",
            "avgRating": 1,
            "count": 1
        }
    }
]

result_top10 = db.ratings.aggregate(pipeline_top10_movies)

for doc in result_top10:
    print(doc)
```
