```python
#A 2dsphere geospatial index in MongoDB is a special type of index that enables efficient querying of geographical data stored in GeoJSON format. 
#It supports queries for geospatial shapes on a spherical surface, such as points, lines, and polygons. 
#This index is ideal for performing queries like finding nearby locations or determining whether a location lies within a specific region.
#collection.create_index([("address.location", "2dsphere")])
# Geospatial Query: Find Properties within a Specific Distance of a Geospatial Point
# Replace the coordinates and maxDistance as needed
geospatial_query = collection.find({
    "address.location": {
        "$near": {
            "$geometry": {"type": "Point", "coordinates": [-8.61308, 41.1413]},
            "$maxDistance": 1000  # Maximum distance in meters
        }
    }
},{"property_type":1,"minimum_nights":1,"host.host_name":1,"maximum_nights":1,"address.location":1,"_id": False}).limit(10)
for document in geospatial_query:
    print(document)
```

```python
#1. Group and Count by Property Type:
#This aggregation pipeline groups the documents by the "property_type" field and counts the number of listings for each property type.

pipeline = [
    {"$group": {"_id": "$property_type", "count": {"$sum": 1}}},
]
result = list(collection.aggregate(pipeline))
for document in result:
    print(document)
```

```python
pipeline = [
    {"$sort": {"price": 1}},  
    {"$limit": 10},  
]
result = list(collection.aggregate(pipeline))
for document in result:
    print(document)
```

```python
#unwind:Converts array elements into individual documents for processing.
pipeline = [
    {"$unwind": "$amenities"},
    {"$group": {"_id": "$amenities", "count": {"$sum": 1}}},
]
result = list(collection.aggregate(pipeline))
for document in result:
    print(document)
```

```python
#5. Bucket: Groups documents into fixed ranges (buckets) based on a field’s value. (Group Documents Based on Price Range:)
#This pipeline groups documents into price range buckets (e.g., low, medium, high) based on the "price" field.

pipeline = [
    {
        "$bucket": {
            "groupBy": "$price",
            "boundaries": [0, 100, 200, 300],  # Specify price range boundaries
            "default": "Other",
            "output": {
                "count": {"$sum": 1},
                "avgPrice": {"$avg": "$price"},
            },
        }
    },
]
result = list(collection.aggregate(pipeline))
for document in result:
    print(document)
```
