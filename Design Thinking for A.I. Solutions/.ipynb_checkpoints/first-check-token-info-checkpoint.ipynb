1. 讀取資料

```python
import pandas as pd
import os

DIR = '/kaggle/input/pump-fun-graduation-february-2025'

train = pd.read_csv(os.path.join(DIR, 'train.csv'))
test = pd.read_csv(os.path.join(DIR, 'test_unlabeled.csv'))

token_info = pd.read_csv(os.path.join(DIR, 'dune_token_info_v2.csv'))
onchain_divers_token_info = pd.read_csv(os.path.join(DIR, 'token_info_onchain_divers_v2.csv'))
```

2. 整理資料
合併 token_info 和 onchain_token_info，讓每個mint有更多特徵：

```python
# 合併兩個資料集
token_info = token_info.rename(columns={'token_mint_address': 'mint'})
onchain_divers_token_info = onchain_divers_token_info.rename(columns={'mint': 'mint'})

token_info_all = pd.merge(token_info, onchain_divers_token_info, on='mint', how='outer')

# Train/Test 補上特徵
train = pd.merge(train, token_info_all, on='mint', how='left')
test = pd.merge(test, token_info_all, on='mint', how='left')
```

3. 前處理
處理一些無法直接丟進模型的地方，例如：

把字串類型的特徵做 Label Encoding

補上缺失值

```python
from sklearn.preprocessing import LabelEncoder

# 對類別型特徵做Label Encoding
cat_cols = train.select_dtypes(include='object').columns.tolist()
cat_cols = [col for col in cat_cols if col != 'mint']

for col in cat_cols:
    le = LabelEncoder()
    all_data = pd.concat([train[col], test[col]], axis=0)
    le.fit(all_data.fillna('NAN'))
    train[col] = le.transform(train[col].fillna('NAN'))
    test[col] = le.transform(test[col].fillna('NAN'))

# 補上其他數值特徵的缺失值
train = train.fillna(-1)
test = test.fillna(-1)
```

4. 建立模型
這題很多人用 CatBoost，因為它對於類別型資料非常友好。

```python
from catboost import CatBoostClassifier

# 特徵
features = [col for col in train.columns if col not in ['mint', 'has_graduated', 'slot_min', 'slot_graduated']]

# 分開X, y
X_train = train[features]
y_train = train['has_graduated']
X_test = test[features]

# 建立模型
model = CatBoostClassifier(
    iterations=2000,
    learning_rate=0.05,
    depth=6,
    eval_metric='Logloss',
    verbose=100,
    random_state=42
)

# 訓練
model.fit(X_train, y_train)
```

5. 預測

```python
# 產生預測機率
preds = model.predict_proba(X_test)[:, 1]

# 輸出提交檔案
submission = pd.DataFrame({
    'mint': test['mint'],
    'has_graduated': preds
})

submission.to_csv('submission.csv', index=False)
```
