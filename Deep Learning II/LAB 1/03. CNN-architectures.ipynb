```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras import Model, layers

(x_train, y_train), (x_test, y_test) = mnist.load_data()
# Normalize pixel values to be between 0 and 1
x_train, x_test = x_train / 255.0, x_test / 255.0

# Reshape the data for CNN input 
x_train = x_train.reshape(x_train.shape[0], 28, 28, 1)
x_test = x_test.reshape(x_test.shape[0], 28, 28, 1)
```

## History
### LeNet-5
- LeNet-5 was developed by Yann LeCun and his colleagues in 1998. It was one of the earliest CNNs and was successfully used for digit recognition tasks, particularly for reading zip codes, digits, etc. 
- It was a 7-layer network with 3 convolutional layers, 2 subsampling layers, and 2 fully connected layers. The activation function used was the sigmoid function.

```python
import tensorflow as tf
from tensorflow.keras import layers, models

def LeNet5(input_shape=(28, 28, 1), num_classes=10):
    # Initialize a sequential model
    model = models.Sequential([
        # First convolutional layer with 6 filters of size 5x5, 'tanh' activation, and same padding
        layers.Conv2D(6, kernel_size=(5, 5), activation='relu', input_shape=input_shape, padding='same'),
        
        # First average pooling layer with 2x2 pool size, reducing spatial dimensions by half
        layers.AveragePooling2D(pool_size=(2, 2)),
        
        # Second convolutional layer with 16 filters of size 5x5, 'relu' activation
        layers.Conv2D(16, kernel_size=(5, 5), activation='tanh'),
        
        # Second average pooling layer with 2x2 pool size, further reducing spatial dimensions
        layers.AveragePooling2D(pool_size=(2, 2)),
        
        # Third convolutional layer with 120 filters of size 5x5, 'tanh' activation
        # Note: No padding, so this reduces the spatial dimensions significantly
        layers.Conv2D(120, kernel_size=(5, 5), activation='tanh'),
        
        # Flatten the output from the last convolutional layer to feed into fully connected layers
        layers.Flatten(),
        
        # Fully connected layer with 84 units and 'relu' activation
        layers.Dense(84, activation='relu'),
        
        # Output layer with units equal to the number of classes, using 'softmax' for probability distribution
        layers.Dense(num_classes, activation='softmax')
    ])
    
    # Return the constructed model
    return model

# Create an instance of the LeNet-5 model
lenet_model = LeNet5()

# Compile the model with the Adam optimizer and sparse categorical cross-entropy loss for multi-class classification
lenet_model.compile(optimizer='adam',
                    loss='sparse_categorical_crossentropy',
                    metrics=['accuracy'])

# Print the model architecture summary to verify the structure
lenet_model.summary()
```

```python

# Train the model
lenet_model.fit(x_train, y_train, epochs=5, validation_data=(x_test, y_test))
```

# The dawn of modern cnns: AlexNet
- AlexNet was developed by Alex Krizhevsky, Ilya Sutskever, and Geoffrey Hinton in 2012. It was the winner of the ImageNet Large Scale Visual Recognition Challenge (ILSVRC) in 2012.
- It was a deep CNN with 8 layers, 5 convolutional layers, and 3 fully connected layers. It used the ReLU activation function and dropout regularization.
- It used data augmentation and dropout to prevent overfitting.
- It used the softmax function in the output layer for classification.



```python
#Important note: The real alexnet used a input shape of 227x227 works with 3 channels, but mnist uses 28x28 with 1 channel (greyscale)

def AlexNet(input_shape=(28, 28, 1), num_classes=10):
    model = tf.keras.models.Sequential([
        # First Convolutional Layer
        tf.keras.layers.Conv2D(32, kernel_size=(3,3), strides=1, padding='same', activation='relu', input_shape=input_shape),
        tf.keras.layers.MaxPooling2D(pool_size=(2,2), strides=2),
        
        # Second Convolutional Layer
        tf.keras.layers.Conv2D(64, kernel_size=(3,3), padding='same', activation='relu'),
        tf.keras.layers.MaxPooling2D(pool_size=(2,2), strides=2),
        
        # Third Convolutional Layer
        tf.keras.layers.Conv2D(128, kernel_size=(3,3), padding='same', activation='relu'),
        
        # Fourth Convolutional Layer
        tf.keras.layers.Conv2D(256, kernel_size=(3,3), padding='same', activation='relu'),
        
        # Fifth Convolutional Layer
        tf.keras.layers.Conv2D(256, kernel_size=(3,3), padding='same', activation='relu'),
        tf.keras.layers.MaxPooling2D(pool_size=(2,2), strides=2),
        
        # Flatten layer
        tf.keras.layers.Flatten(),
        
        # Fully connected layers
        tf.keras.layers.Dense(1024, activation='relu'),
        tf.keras.layers.Dropout(0.5),
        tf.keras.layers.Dense(512, activation='relu'),
        tf.keras.layers.Dropout(0.5),
        
        # Output layer
        tf.keras.layers.Dense(num_classes, activation='softmax')
    ])
    
    return model

# Create AlexNet model
alexnet_model = AlexNet(input_shape=(28, 28, 1), num_classes=10)

# Compile the model
alexnet_model.compile(optimizer='adam',
                      loss='sparse_categorical_crossentropy',
                      metrics=['accuracy'])

# Print model summary
alexnet_model.summary()

# Fit the model
alexnet_model.fit(x_train, y_train, epochs=5, validation_data=(x_test, y_test))
```

```python
from sklearn.metrics import confusion_matrix
import numpy as np

# Get predictions
predictions = alexnet_model.predict(x_test)
predicted_labels = np.argmax(predictions, axis=1)

# Compute confusion matrix
cm = confusion_matrix(y_test, predicted_labels)
# Print confusion matrix
print("Confusion Matrix:")
print("----------------")
for i in range(10):
    for j in range(10):
        print(f"{cm[i][j]:4d}", end=" ")
    print()

# Calculate and print accuracy
accuracy = np.sum(np.diag(cm)) / np.sum(cm)
print(f"\nAccuracy: {accuracy:.4f}")

# Print some additional metrics
print("\nPer-class accuracy:")
for i in range(10):
    class_accuracy = cm[i][i] / np.sum(cm[i])
    print(f"Class {i}: {class_accuracy:.4f}")
```
