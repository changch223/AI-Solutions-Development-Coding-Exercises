## YOLO → ImageFolder 

```python
import os
import shutil
from pathlib import Path
import yaml

# ==== 路徑設置 ====
yolo_root = Path("C:/Users/wayne/OneDrive/桌面/Final Project Dataset/yolo")
output_root = Path("C:/Users/wayne/OneDrive/桌面/Final Project Dataset/imagefolder")

splits = ['train', 'valid', 'test']

# ==== 讀取 data.yaml 取得 class 名稱 ====
yaml_path = yolo_root / "data.yaml"
with open(yaml_path, 'r') as f:
    yolo_config = yaml.safe_load(f)

class_names = yolo_config['names']  # e.g., ['biodegradable', 'cardboard', 'metal']

# ==== 開始轉換 ====
for split in splits:
    print(f"Processing split: {split}")
    image_dir = yolo_root / split / "images"
    label_dir = yolo_root / split / "labels"

    for label_file in label_dir.glob("*.txt"):
        # 取得圖像檔案名稱
        stem = label_file.stem  # 不含副檔名的檔名
        image_path_jpg = image_dir / f"{stem}.jpg"
        image_path_png = image_dir / f"{stem}.png"
        image_path_jpeg = image_dir / f"{stem}.jpeg"

        # 找到對應圖像
        if image_path_jpg.exists():
            image_path = image_path_jpg
        elif image_path_png.exists():
            image_path = image_path_png
        elif image_path_jpeg.exists():
            image_path = image_path_jpeg
        else:
            print(f"⚠️ Image not found for: {label_file}")
            continue

        # 讀取 label 第一行取得 class index（假設只有一個物體）
        with open(label_file, 'r') as f:
            first_line = f.readline().strip()
        class_id = int(first_line.split()[0])
        class_name = class_names[class_id]

        # 目標資料夾
        target_dir = output_root / split / class_name
        target_dir.mkdir(parents=True, exist_ok=True)

        # 複製圖片
        shutil.copy(image_path, target_dir / image_path.name)

print("✅ YOLOv8 dataset successfully converted to ImageFolder format!")
```

```python
from pathlib import Path
from collections import defaultdict
from sklearn.model_selection import train_test_split
import shutil

# === 設定 ===
original_yolo = Path("C:/Users/wayne/OneDrive/桌面/Final Project Dataset/yolo")
output_yolo = Path("C:/Users/wayne/OneDrive/桌面/Final Project Dataset/yolo_resplit")
splits = ['train', 'valid', 'test']
image_exts = ['.jpg', '.jpeg', '.png']

# 統一收集所有 (image, label, class_id)
all_data = []

for split in splits:
    image_dir = original_yolo / split / "images"
    label_dir = original_yolo / split / "labels"

    for img_path in image_dir.glob("*"):
        if img_path.suffix.lower() not in image_exts:
            continue
        lbl_path = label_dir / (img_path.stem + ".txt")
        if not lbl_path.exists():
            continue
        with open(lbl_path) as f:
            lines = f.readlines()
            if not lines:
                continue
            class_id = lines[0].split()[0]
            all_data.append((img_path, lbl_path, class_id))

# 分類後重新分割
from collections import defaultdict
classwise = defaultdict(list)
for img, lbl, cls in all_data:
    classwise[cls].append((img, lbl))

# 分配並複製
for split in ['train', 'valid', 'test']:
    (output_yolo / split / 'images').mkdir(parents=True, exist_ok=True)
    (output_yolo / split / 'labels').mkdir(parents=True, exist_ok=True)

for cls_id, items in classwise.items():
    items = sorted(items)
    trainval, test = train_test_split(items, test_size=0.15, random_state=42)
    train, valid = train_test_split(trainval, test_size=0.1765, random_state=42)  # 15% of total

    for split, data in zip(['train', 'valid', 'test'], [train, valid, test]):
        for img_path, lbl_path in data:
            dst_img = output_yolo / split / "images" / img_path.name
            dst_lbl = output_yolo / split / "labels" / lbl_path.name
            shutil.copy2(img_path, dst_img)
            shutil.copy2(lbl_path, dst_lbl)

print("✅ 重新分割完成：已複製到 yolo_resplit/ 資料夾")
```

## 重分類腳本（重新分配 train/valid/test，比例 70/15/15）

```python
import os
import shutil
from pathlib import Path
from sklearn.model_selection import train_test_split

# ==== 原始 ImageFolder 路徑 ====
original_data_dir = Path("C:/Users/wayne/OneDrive/桌面/Final Project Dataset/imagefolder")
output_data_dir = Path("C:/Users/wayne/OneDrive/桌面/Final Project Dataset/imagefolder_resplit")

# ==== 分割比例 ====
train_ratio = 0.7
val_ratio = 0.15
test_ratio = 0.15

# ==== 建立目標資料夾結構 ====
splits = ['train', 'valid', 'test']
class_names = [d.name for d in (original_data_dir / "train").iterdir() if d.is_dir()]
for split in splits:
    for class_name in class_names:
        (output_data_dir / split / class_name).mkdir(parents=True, exist_ok=True)

# ==== 對每個類別重分配圖片 ====
for class_name in class_names:
    # 收集所有圖片
    all_imgs = []
    for split in splits:
        img_dir = original_data_dir / split / class_name
        imgs = list(img_dir.glob("*.[jp][pn]g")) + list(img_dir.glob("*.jpeg"))
        all_imgs.extend(imgs)

    # 分割為 train / val / test
    train_imgs, temp_imgs = train_test_split(all_imgs, train_size=train_ratio, random_state=42, shuffle=True)
    val_imgs, test_imgs = train_test_split(temp_imgs, test_size=test_ratio / (val_ratio + test_ratio), random_state=42)

    # 複製圖片
    for img in train_imgs:
        shutil.copy(img, output_data_dir / "train" / class_name / img.name)
    for img in val_imgs:
        shutil.copy(img, output_data_dir / "valid" / class_name / img.name)
    for img in test_imgs:
        shutil.copy(img, output_data_dir / "test" / class_name / img.name)

    print(f"[{class_name}] train={len(train_imgs)}, val={len(val_imgs)}, test={len(test_imgs)}")

print("✅ 重分類完成！資料已重新分配到 imagefolder_resplit/")
```

## YOLO ➜ COCO JSON

```python
import os
import json
from pathlib import Path
from PIL import Image

# === 參數設定 ===
yolo_root = Path("C:/Users/wayne/OneDrive/桌面/Final Project Dataset/yolo_resplit/train")  # 改為 train/valid/test 路徑
image_dir = yolo_root / "images"
label_dir = yolo_root / "labels"
output_json_path = yolo_root / "annotations_coco.json"

# 手動指定類別名稱（順序需與 class_id 對應）
categories_list = [
    "BIODEGRADABLE",
    "CARDBOARD",
    "GLASS",
    "METAL",
    "PAPER",
    "PLASTIC"
]

# === 產生 COCO 結構 ===
coco_dict = {
    "images": [],
    "annotations": [],
    "categories": []
}

for i, category_name in enumerate(categories_list):
    coco_dict["categories"].append({
        "id": i,
        "name": category_name,
        "supercategory": "trash"
    })

annotation_id = 1

for image_id, image_file in enumerate(sorted(image_dir.glob("*"))):
    if image_file.suffix.lower() not in [".jpg", ".jpeg", ".png"]:
        continue

    # 讀取圖片尺寸
    with Image.open(image_file) as img:
        width, height = img.size

    # 加入 image metadata
    coco_dict["images"].append({
        "id": image_id,
        "file_name": image_file.name,
        "width": width,
        "height": height
    })

    # 找對應標註
    label_path = label_dir / (image_file.stem + ".txt")
    if not label_path.exists():
        continue

    with open(label_path, 'r') as f:
        for line in f.readlines():
            parts = line.strip().split()
            if len(parts) != 5:
                continue

            class_id = int(parts[0])
            x_center, y_center, w, h = map(float, parts[1:])
            # 轉換為左上角 (x,y,w,h)
            x = (x_center - w / 2) * width
            y = (y_center - h / 2) * height
            w *= width
            h *= height

            # 加入 annotation
            coco_dict["annotations"].append({
                "id": annotation_id,
                "image_id": image_id,
                "category_id": class_id,
                "bbox": [x, y, w, h],
                "area": w * h,
                "iscrowd": 0
            })
            annotation_id += 1

# === 輸出 JSON ===
with open(output_json_path, 'w') as f:
    json.dump(coco_dict, f, indent=2)

print(f"✅ COCO JSON saved to {output_json_path}")
```
